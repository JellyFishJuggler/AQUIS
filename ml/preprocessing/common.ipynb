```python
import pandas as py
import numpy as np
```

```python
data = py.read_json('../data/data/raw/groundwater.jsonl',lines=True)
df = py.DataFrame(data)
print(df.head())
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Groundwater Level Telemetry 6 Hourly (meter)'].describe()
```

```python
df['Station'].nunique()
```

```python
df['Station'].value_counts().head(20)
```

```python
df['Data Acquisition Time'] = py.to_datetime(df['Data Acquisition Time'])
```

```python
level_col = "Groundwater Level Telemetry 6 Hourly (meter)"

df[level_col] = py.to_numeric(
    df[level_col],
    errors="coerce"
)
```

```python
df.duplicated().sum()
```

```python
df.duplicated(
    subset=["Station", "Data Acquisition Time"]
).sum()
```

```python
df = df.sort_values(
    ["Station", "Data Acquisition Time"]
).reset_index(drop=True)
```

```python
df.head()
```

```python
df_path = '../data/processed/common.parquet'
df.to_parquet(
    df_path,
    index=False
)
print(f'Saved to {df_path}')
```
